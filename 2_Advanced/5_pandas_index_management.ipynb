```python
# Import libraries'

import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading the Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data cleanup 
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
median_pivot = df_usa.pivot_table(values='salary_year_avg',index='job_title_short',aggfunc=['median','min','max'])
```

```python
median_pivot
```

```python
median_pivot.sort_values(by=('median','salary_year_avg'), ascending=False)
```

```python
median_pivot.plot(kind='bar',figsize=(12,6),title='Median Salary by Job Title') 
```
